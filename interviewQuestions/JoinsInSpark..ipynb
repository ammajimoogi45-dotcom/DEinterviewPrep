##Broadcast hash Join
- when one of the two dataframe can be store in the spark memory entirely
- spark automatically does a  broadcase hash join.
- spark check if any of the dataframes are within broadcastTreshold or not , if within the treshold it will pickup broadcast hash join.
- spark.sql.autoBroadcastJoinThreshold()
- broadcast mode is used to inform how broadcastside is packaged before it send to the exectuors over network.
> hashrelation broadcast mode :
- where the data is stored in hash map ( key value pair) format and send to the executors. , in this mode the key is stored inside the list.
- steps include , broadcast, hash( build), probe 

##Shuffle hash join
- when one of the dataframes can't be store entirely in the memory
- after shuffled the data over the network , when we are able to hash the partitions of the executors , then this join is picked.
> steps :
-     > shuffle : shuffle both the dataframes over the network. hash(joinKey)% SparkPartition
-     > build : builds the hash map for smaller partition in the executor. joinKey - value.
-     > probe : will do a look up in the hash map using the  larger table key 

##Broadcast Nested Join
- when we join two dataframes using the non-equi conditions ( >=, <=,>,...)
- one dataset is small enough to broadcast.
- cross join 
- Disjunctive predicates: Conditions involving OR can lead to BNLJ when Spark cannot extract usable equi-join keys.
- Subqueries: Certain rewritten subquery plans can use nested-loop joins.
- Anti-joins: Some anti-join plans can use BNLJ, but not every anti-join requires it.
> step:
> - broadcast the smaller dataset
> - applies a nested loop check for each row in one dataframe with each row in the other dataframe.

###Coalesce vs Repartition 
- coalesce : it reduces the partitions without any shuffle . ( no increase in partition is possible.)
- Repartition : it increases or reduces the partitions with full shuffle
